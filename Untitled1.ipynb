### 5. Clean the data

After the initial EDA, the data will be thoroughly cleaned to ensure its quality and suitability for machine learning models. This step involves several crucial operations:

*   **Handling Missing Values:**
    *   **Identification:** Confirm missing values identified during EDA using `.isnull().sum()`.
    *   **Strategy:** Decide on a strategy based on the nature and quantity of missing data:
        *   **Imputation:** For numerical data, use mean, median, or mode imputation. For time-series data, forward-fill, backward-fill, or interpolation methods can be effective. For categorical data, mode imputation or assigning a new category 'Unknown' can be considered.
        *   **Removal:** If a column has a very high percentage of missing values (e.g., >70-80%), or if specific rows have critical missing information that cannot be reliably imputed, those columns or rows might be removed. This should be done cautiously.

*   **Correcting Errors:**
    *   **Data Type Conversion:** Ensure all columns have appropriate data types (e.g., convert 'object' columns containing numbers to numeric types, date strings to datetime objects).
    *   **Inconsistent Data Entries:** Standardize categorical values (e.g., 'USA', 'U.S.A.', 'United States' to 'USA'). This often involves string manipulation methods (e.g., `.str.lower()`, `.str.replace()`, `.map()`).
    *   **Logical Errors:** Address any values that are logically impossible or incorrect based on domain knowledge (e.g., negative quantities, delivery times exceeding reasonable limits).

*   **Removing Duplicates:**
    *   **Identify Duplicates:** Use `.duplicated().sum()` to count duplicate rows.
    *   **Remove Duplicates:** Remove identified duplicate rows using `.drop_duplicates(inplace=True)` to ensure that each record is unique and does not skew analysis or model training.

*   **Addressing Outliers:**
    *   **Review and Validate:** Re-examine outliers identified during EDA. Determine if they are genuine extreme values or data entry errors.
    *   **Handling Strategies:**
        *   **Capping/Winsorization:** For numerical outliers that are genuine but highly influential, cap them at a certain percentile (e.g., 99th percentile) to reduce their impact.
        *   **Transformation:** Apply mathematical transformations (e.g., logarithmic, square root) to skewed distributions to reduce the impact of outliers and make the data more normally distributed.
        *   **Removal:** In rare cases, if an outlier is clearly an error and significantly distorts the data, it might be removed. This decision should be well-justified.

### 6. Preprocess the data

After cleaning, the data will be preprocessed to transform it into a format suitable for machine learning algorithms. This involves several key steps:

*   **Data Type Conversion:**
    *   Ensure all columns are of the correct data type. This is crucial for numerical operations and proper handling of dates/times. For example, convert relevant columns to `datetime` objects using `pd.to_datetime()` and numerical columns to appropriate numeric types (e.g., `int`, `float`).

*   **Feature Engineering:** Create new, meaningful features from existing ones to enhance the predictive power of models.
    *   **Time-Series Features:** For demand and routing data, extract temporal features such as year, month, day of week, day of month, hour, minute, week of year, quarter, and holidays. Lag features (e.g., demand from previous days/weeks) and rolling window statistics (e.g., moving averages of demand, average travel time over the last hour) will be created.
    *   **Geospatial Features:** Calculate distances between locations, cluster locations, or extract features based on geographical characteristics.
    *   **Interaction Features:** Combine existing features (e.g., `price * quantity`, `day_of_week * is_holiday`).
    *   **Domain-Specific Features:** Create features based on business knowledge (e.g., `time_since_last_order`, `promotion_duration`).

*   **Handling Categorical Variables:**
    *   **One-Hot Encoding:** Convert nominal categorical variables (e.g., 'product_category', 'delivery_zone') into a numerical format using one-hot encoding (e.g., `pd.get_dummies()`).
    *   **Label Encoding/Ordinal Encoding:** For ordinal categorical variables (e.g., 'delivery_priority': Low, Medium, High), convert them into numerical representations respecting their order.
    *   **Target Encoding/Frequency Encoding:** For high-cardinality categorical features, consider more advanced encoding techniques.

*   **Scaling Numerical Features:** Standardize or normalize numerical features to ensure that no single feature dominates the learning process due to its scale.
    *   **Standardization (Z-score normalization):** Use `StandardScaler` to transform data to have a mean of 0 and a standard deviation of 1. This is generally preferred for algorithms that assume normally distributed data (e.g., linear regression, SVMs).
    *   **Normalization (Min-Max scaling):** Use `MinMaxScaler` to scale features to a fixed range, usually between 0 and 1. This is useful for algorithms that are not sensitive to normal distribution assumptions (e.g., neural networks).

*   **Time Series Specific Preprocessing:**
    *   **Time Indexing:** Ensure the time-series data is correctly indexed by datetime objects.
    *   **Resampling:** Aggregate or disaggregate data to a consistent frequency if needed (e.g., hourly to daily, daily to weekly).
    *   **Handling Seasonality and Trend:** While models can handle these, sometimes explicit decomposition or differencing might be performed depending on the chosen forecasting model.

### 2. Split the unified dataset into training, validation, and test sets

For time-series demand forecasting, it is critical to perform a time-based split of the data to ensure that the model is trained on past data and evaluated on future data. This prevents data leakage and provides a more realistic assessment of the model's predictive performance.

#### Strategy for Time-Based Split:

1.  **Sort Data:** Ensure the unified dataset (`df_unified`) is sorted chronologically by the `TIME_COLUMN` (e.g., `date`).
2.  **Define Split Points:** Determine specific cutoff dates or percentages of the data to create the splits.
    *   **Training Set:** The earliest data, used to train the model.
    *   **Validation Set:** Data immediately following the training set, used for hyperparameter tuning and model selection.
    *   **Test Set:** The most recent data, held back until the final model evaluation to simulate real-world performance.

#### Example Conceptual Splitting (assuming `df_unified` exists):

```python

### 3. Select appropriate machine learning algorithms for demand prediction

The choice of machine learning algorithms for demand prediction depends on the nature of the data, the desired prediction horizon, the complexity of the patterns, and interpretability requirements. We will consider both traditional time-series models and more general regression-based machine learning models, as time-series data often contains features that can be effectively modeled by supervised learning algorithms.

#### A. Time-Series Specific Models:
These models are designed to capture temporal dependencies, trends, and seasonality inherently.

*   **ARIMA (AutoRegressive Integrated Moving Average) / SARIMA (Seasonal ARIMA):** Effective for stationary or stationarizable time series with identifiable trends and seasonal patterns. Requires careful differencing and parameter tuning.
*   **Exponential Smoothing (ETS, Holt-Winters):** Good for data with trend and/or seasonality. Simpler to implement and often provides strong baselines.
*   **Prophet (from Facebook):** Robust to missing data, outliers, and able to handle multiple seasonality periods (e.g., daily, weekly, yearly). It is particularly useful for business forecasting due to its intuitive parameters.

#### B. Regression-Based Machine Learning Models:
These models treat the demand forecasting problem as a supervised regression task, leveraging the rich set of features engineered in previous steps (lagged demand, calendar features, external factors, etc.).

*   **Linear Regression:** A simple baseline, useful for understanding linear relationships between features and demand.
*   **Random Forest Regressor:** An ensemble method that can capture non-linear relationships and interactions between features. Less prone to overfitting than decision trees.
*   **Gradient Boosting Regressors (e.g., XGBoost, LightGBM, CatBoost):** Highly powerful and popular algorithms for tabular data. They build an ensemble of weak prediction models (typically decision trees) sequentially. They are known for high accuracy and can handle complex patterns, feature interactions, and various data types efficiently.
*   **Support Vector Regressor (SVR):** Can be effective for non-linear relationships, especially with appropriate kernel selection.
*   **Neural Networks (e.g., LSTMs, Feed-forward NNs):** For highly complex patterns or large datasets, especially if sequences or deep interactions are critical. LSTMs (Long Short-Term Memory) are particularly suited for sequence data like time series.

#### Selection Strategy:

*   **Start with Baselines:** Begin with simpler models (e.g., Exponential Smoothing, Linear Regression) to establish a baseline performance.
*   **Explore Advanced Models:** Progress to more sophisticated models (e.g., Prophet, XGBoost, Random Forest) that can capture complex patterns and leverage the engineered features.
*   **Consider Ensembling:** Combine predictions from multiple models to potentially achieve better performance and robustness.
*   **Focus on Business Needs:** The final choice will also depend on factors like prediction accuracy requirements, computational resources, and the need for interpretability.

In the initial phase, a few promising algorithms (e.g., Prophet for its robustness and XGBoost/LightGBM for their performance with rich features) will be selected for training and evaluation.

#### B. Evaluation Metrics:

We will use standard regression metrics to quantify prediction accuracy:

*   **Mean Absolute Percentage Error (MAPE):** Average of the absolute percentage errors. Useful for understanding error in relative terms, but can be problematic with zero or near-zero actual values.

## Define the Dynamic Routing Problem and Select Algorithms

### Subtask:
Clearly define the dynamic routing problem, including its objectives, constraints, and inputs. Based on this definition, select appropriate algorithms or approaches for solving the Vehicle Routing Problem (VRP) or its dynamic variants.


### 1. Outline the key objectives of the dynamic routing algorithm

The primary objectives for the dynamic routing algorithm are centered around optimizing efficiency, cost-effectiveness, and customer satisfaction in logistics operations. These include:

*   **Minimizing Total Travel Time:** Reducing the overall time spent by vehicles on routes, which directly impacts delivery speed and operational efficiency.
*   **Minimizing Total Travel Distance:** Decreasing the cumulative distance covered by the fleet, leading to reduced fuel consumption and vehicle wear.
*   **Minimizing Operational Costs:** This encompasses fuel costs, driver wages (by optimizing working hours), and vehicle maintenance costs.
*   **Maximizing Deliveries/Service Points:** Ensuring that as many delivery or service requests as possible are fulfilled within operational constraints, especially important in high-demand periods.
*   **Meeting Time Window Constraints:** Adhering to specified delivery or pickup time windows for customers, which is critical for customer satisfaction and service level agreements.
*   **Balancing Vehicle Utilization:** Efficiently distributing workload across the fleet to avoid overloading some vehicles while others are underutilized.
*   **Minimizing Late Deliveries/Customer Wait Times:** Improving customer experience by ensuring timely deliveries and reducing uncertainty.
*   **Adapting to Dynamic Changes:** The algorithm must be robust enough to handle real-time changes such as new orders, cancellations, traffic incidents, or vehicle breakdowns with minimal disruption.

### 2. List the primary constraints that the algorithm must consider

To ensure the dynamic routing algorithm generates feasible and realistic routes, it must operate within a set of well-defined constraints. These constraints reflect real-world limitations and operational rules:

*   **Vehicle Capacity Constraints:** Each vehicle has a maximum weight and/or volume it can carry. The total load for any route must not exceed the vehicle's capacity.
*   **Driver Working Hours and Rest Periods:** Routes must comply with legal and company-specific regulations regarding driver working hours, breaks, and rest periods to ensure safety and prevent fatigue.
*   **Delivery/Pickup Time Window Constraints:** Many customer orders or service points have specific time windows during which delivery or pickup must occur. The algorithm must ensure these windows are met.
*   **Fleet Size and Availability:** The routing solutions must be constrained by the number of available vehicles and their operational status (e.g., in-maintenance, available).
*   **Vehicle Speed Limits and Road Restrictions:** Routes must account for legal speed limits, road types, one-way streets, bridge height/weight restrictions, and other geographical limitations.
*   **Traffic Conditions:** Real-time and predicted traffic patterns must be considered to estimate accurate travel times and avoid congested areas where possible.
*   **Depot/Start/End Locations:** Vehicles typically start and end their routes at designated depots or specific operational hubs.
*   **Service Time at Locations:** The time required to perform the delivery or service at each customer location must be factored into route calculations.
*   **Customer Prioritization:** Some deliveries might have higher priority than others and must be fulfilled sooner or with greater certainty.
*   **Fuel Range and Refueling Stops:** For longer routes, the fuel capacity of vehicles and the need for potential refueling stops might act as a constraint.
*   **Homogeneous/Heterogeneous Fleet:** The algorithm must distinguish between different vehicle types with varying capacities, speeds, or specialized equipment (e.g., refrigerated trucks).

### 3. Identify the main inputs required for the algorithm

To power the dynamic routing algorithm and enable it to make informed decisions, a variety of data inputs are essential. These inputs can be static or dynamic/real-time:

#### Static/Pre-defined Inputs:
*   **Customer Locations/Delivery Points:** Geographic coordinates (latitude, longitude) for all potential delivery or pickup locations, along with any specific delivery instructions or access restrictions.
*   **Depot Locations:** Geographic coordinates of all starting and ending points for vehicles.
*   **Vehicle Fleet Information:** Details for each vehicle, including:
    *   Capacity (weight and/or volume).
    *   Vehicle type (e.g., van, truck, refrigerated).
    *   Operating costs per km/hour.
    *   Fuel efficiency/range.
*   **Road Network Data:** Digital maps containing information on roads, speed limits, one-way streets, turns restrictions, and distance/travel time matrices between locations.
*   **Service Times:** Average or estimated time required to perform service at each customer location.
*   **Driver Shift Schedules:** Pre-defined working hours, breaks, and availability for each driver.

#### Dynamic/Real-time Inputs (often requiring integration with external systems or sensors):
*   **Predicted Demand/Order Information:** From the demand forecasting model, including:
    *   New customer orders (location, quantity, time window, priority).
    *   Order cancellations or modifications.
    *   Predicted demand for specific regions or time periods (to proactively position vehicles).
*   **Real-time Traffic Data:** Current and predicted traffic conditions (e.g., congestion, incidents, road closures) influencing travel times.
*   **Real-time Vehicle Status:** Current location (GPS data), speed, remaining capacity, estimated time of arrival (ETA) at next stop, and availability (e.g., currently busy, available).
*   **Weather Data:** Current and forecasted weather conditions that might affect travel times or road accessibility.
*   **Driver Availability Updates:** Real-time changes to driver status (e.g., unexpected illness, early finish).
*   **External Events:** Information on public holidays, large events, or emergencies that might impact logistics operations.

### 4. Select suitable dynamic routing algorithms or heuristics

Given the dynamic nature of the problem, the need for real-time adaptability, and the potential for large-scale operations, a combination of algorithmic approaches is generally most effective. Pure exact methods are often computationally intractable for real-world dynamic VRPs due to their NP-hard complexity.

#### Recommended Approaches:

1.  **Metaheuristics (Primary Approach):**
    *   **Description:** Metaheuristics are approximation algorithms designed to find good-enough solutions to hard optimization problems in a reasonable amount of time. They are particularly well-suited for large-scale and dynamic VRPs where optimality is often sacrificed for speed and flexibility.
    *   **Specific Algorithms:**
        *   **Genetic Algorithms (GAs):** Inspired by natural selection, GAs can explore a large search space effectively. They are robust to changes and can be adapted to handle various VRP variants and constraints. Their parallelizability can also be leveraged for faster execution.
        *   **Simulated Annealing (SA):** An algorithm that can escape local optima by allowing 'bad' moves with a certain probability, which decreases over time. It's effective for complex search spaces and can be adapted for dynamic scenarios.
        *   **Tabu Search (TS):** A local search algorithm that uses a "tabu list" to avoid revisiting recently explored solutions, encouraging exploration of new regions of the search space. It's known for its ability to produce high-quality solutions for VRPs.
        *   **Ant Colony Optimization (ACO):** Inspired by the foraging behavior of ants, ACO can find good paths in graphs and is suitable for VRPs, especially when dealing with complex constraints.
    *   **Justification:** Metaheuristics offer a good balance between solution quality and computational time, making them suitable for dynamic re-optimization. They can handle various constraints and integrate real-time updates. Their flexibility allows for customization to specific business rules and objectives.

2.  **Hybrid Approaches (Combining Metaheuristics with Local Search or Machine Learning):**
    *   **Description:** Often, combining a metaheuristic with a more refined local search (e.g., 2-opt, 3-opt, exchange procedures) can significantly improve solution quality. Additionally, Machine Learning models can be used to predict traffic, service times, or even guide the search process of the metaheuristic.
    *   **Justification:** This approach leverages the global exploration capabilities of metaheuristics with the local exploitation power of local search, leading to better overall solutions. Integrating ML allows for more intelligent decision-making based on learned patterns.

3.  **Heuristics for Dynamic Problem Handling:**
    *   **Description:** Specific heuristics designed to quickly re-optimize routes when new events (new orders, cancellations, traffic changes) occur. These often involve re-inserting new requests into existing routes or performing partial re-optimizations.
    *   **Specific Heuristics:** Insertion heuristics (e.g., cheapest insertion), sweep algorithm, or adaptive large neighborhood search (ALNS).
    *   **Justification:** For dynamic scenarios, re-running a full metaheuristic from scratch for every minor change might be too slow. Heuristics provide quick, reactive adjustments to maintain efficiency.

#### Considerations for Implementation:

*   **Scalability:** Metaheuristics and heuristics are generally more scalable to large numbers of vehicles and delivery points compared to exact methods.
*   **Computational Complexity:** These methods provide good solutions within practical time limits, crucial for dynamic adjustments.
*   **Effectiveness for Dynamic Scenarios:** The ability to quickly incorporate new information and re-optimize routes is paramount. Hybrid and dedicated dynamic heuristics, guided by metaheuristics, are key.

For a practical system, the core would likely involve a metaheuristic that runs periodically or when significant changes occur, complemented by faster heuristics for immediate, minor adjustments and potentially an ML component for predictive insights.

### 2. Model the physical environment and road network

To simulate vehicle movements and delivery operations accurately, a robust representation of the physical environment and the underlying road network is essential. This step outlines how to model these aspects.

#### Geographical Locations:
*   **Coordinates:** All key locations (depots, customer delivery points) will be represented by their precise geographic coordinates (latitude and longitude). This allows for realistic distance calculations and mapping.
*   **Geocoding:** If raw addresses are provided, a geocoding service (e.g., Google Maps Geocoding API, OpenStreetMap Nominatim) will be used to convert them into latitude and longitude coordinates.

#### Road Network Representation:

*   **Simplified Grid (for initial prototyping/testing):**
    *   **Description:** For simpler simulations or initial testing phases, the environment can be conceptualized as a grid where vehicles move between discrete points. Distances are calculated using Manhattan distance or Euclidean distance.
    *   **Pros:** Easy to implement, computationally inexpensive.
    *   **Cons:** Less realistic, does not account for actual road topology, traffic, or one-way streets.

*   **Graph-based Representation (for realistic simulations):**
    *   **Description:** The logistics network will be modeled as a graph where intersections and key locations are represented as **nodes**, and roads connecting them are represented as **edges**.
    *   **Nodes:** Geographic coordinates, unique ID.
    *   **Edges:** Represent road segments between nodes.
        *   **Attributes:**
            *   **Distance:** Actual physical length of the road segment.
            *   **Base Travel Time:** Travel time under free-flow conditions (calculated from distance and speed limit).
            *   **Speed Limits:** Associated speed limits for each road segment.
            *   **Road Type:** (e.g., highway, urban road, residential street) which can influence speed and traffic patterns.
            *   **One-Way Restrictions:** Indicate if a road segment is one-way.
            *   **Traffic Conditions:** Capacity or congestion levels (for dynamic updates).

*   **Distance and Travel Time Calculation:**
    *   **Matrix Generation:** A pre-computed distance and travel time matrix between all relevant points (depots and customer locations) can be generated using a routing engine (e.g., Google Maps Directions API, OSRM, Valhalla, OpenRouteService). This matrix will store: `Distance(A, B)` and `TravelTime(A, B)`.
    *   **Dynamic Updates:** For more realism, the travel times within the graph or matrix will be dynamically updated based on real-time traffic data (as discussed in the 'Dynamic Elements' section).

This structured representation of the physical environment allows the simulation to accurately model vehicle movement, calculate realistic travel times, and provide a foundation for routing algorithms.

### 3. Implement dynamic elements for the simulation

To create a realistic and responsive simulation environment, several elements must be modeled dynamically. These elements will change over time and influence the overall state and outcome of the simulation.

#### Vehicles:
*   **Dynamic State:** Vehicles will have an evolving state that includes their `current_location`, `current_load` (weight/volume), `remaining_capacity`, `current_route_segment`, and `estimated_time_of_arrival` (ETA) at the next stop.
*   **Movement Logic:** Implement logic for vehicles to move between nodes/locations based on travel times derived from the road network. This involves updating their position at each time step.
*   **Loading/Unloading:** Simulate the process of loading goods at depots and unloading at customer locations, accounting for `service_time`.
*   **Route Re-planning:** Vehicles should be capable of re-planning their routes dynamically in response to new orders, traffic changes, or unexpected events.
*   **Driver Schedules:** Incorporate driver working hours, breaks, and compliance with regulations to ensure realistic vehicle availability and movement.

#### Orders/Demand:
*   **Order Arrival Generation:** New orders will arrive dynamically throughout the simulation. This can be modeled using:
    *   **Historical Data Patterns:** Extracting patterns from historical order data to generate new orders with realistic distributions of location, quantity, and time windows.
    *   **Stochastic Processes:** Using probability distributions (e.g., Poisson process for arrival rate) to simulate random order arrivals.
    *   **Demand Prediction Model Output:** Directly integrating the output from the demand prediction model (e.g., forecasted demand for specific regions or product types for an upcoming hour/day) to generate orders that reflect anticipated future demand.
*   **Order Attributes:** Each new order will be created with its defined attributes: `origin/destination`, `quantity/weight/volume`, `service_time`, `time_window`, and `priority`.

#### Traffic Conditions:
*   **Dynamic Travel Times:** Travel times on road segments will not be static but will change over the simulation period.
*   **Modeling Approaches:**
    *   **Simple Factor-based:** Applying a dynamic multiplier to base travel times based on the time of day (e.g., higher multiplier during peak hours) or random fluctuations.
    *   **Historical Traffic Patterns:** Using historical traffic data to generate typical congestion patterns throughout the day or week.
    *   **Real-time Traffic Integration:** Simulating the ingestion of real-time traffic updates (e.g., from APIs) that modify segment travel times instantly.
    *   **Incident Generation:** Simulating unexpected incidents like accidents or road closures that significantly impact travel times on specific segments.
*   **Impact on Routing:** The routing algorithms within the simulation must be able to react to these dynamic traffic changes by recalculating optimal paths or suggesting alternative routes.

### 4. Establish a time progression mechanism

A critical aspect of any simulation is its ability to progress through time, allowing dynamic elements to evolve and events to unfold. This step outlines the mechanism for advancing the simulation clock and managing events.

#### Time Representation:
*   **Discrete-Event Simulation (DES):** The simulation progresses by jumping from one event time to the next, rather than through fixed time steps. This is efficient when events are sparse.
*   **Time-Stepped Simulation:** The simulation progresses in fixed, regular time increments (e.g., every minute, every hour). This is suitable when continuous updates are needed or events are frequent and regularly spaced.
*   **Hybrid Approach:** A combination of both, where fixed time steps are used for general progression, but specific events trigger immediate updates.

#### Simulation Clock:
*   A central `simulation_time` variable will keep track of the current time within the simulation (e.g., using `datetime` objects or a simple counter).

#### Event Queue (for Discrete-Event Simulations):
*   Maintain a prioritized queue of future events (e.g., `vehicle_arrives_at_location`, `new_order_arrives`, `traffic_update`).
*   At each step, the simulation clock advances to the time of the next event in the queue.
*   When an event is processed, its effects are applied, and new future events might be scheduled.

#### Event Processing Logic:
*   **Order Arrival:** When a new order arrives, it's added to a pool of pending orders, potentially triggering a re-evaluation of routing decisions.
*   **Vehicle Movement:** At each time step (or event), update the vehicle's position based on its speed and current route. Calculate expected arrival times at subsequent stops.
*   **Delivery/Pickup Completion:** When a vehicle arrives at a customer location, simulate the `service_time` and mark the order as delivered/picked up. Update vehicle load and capacity.
*   **Traffic Updates:** Periodically (or event-driven), update travel times on road segments based on dynamic traffic information.
*   **Route Re-evaluation:** Define triggers for when routing algorithms should be run (e.g., new order arrival, significant traffic change, vehicle completion of a segment).

# Conceptual simulation loop
while simulation_time < end_time:
    # Process events at current_simulation_time
    process_current_events(simulation_time)

    # Update vehicle positions and states
    update_vehicle_states(simulation_time)

    # Generate new orders (if applicable)
    generate_new_orders(simulation_time)

    # Apply dynamic traffic changes
    apply_traffic_updates(simulation_time)

    # Check for routing re-optimization needs
    check_and_trigger_routing_optimization(simulation_time)

    # Advance simulation_time (for time-stepped) or jump to next event time (for DES)
    simulation_time = advance_time()
```
This time progression mechanism ensures that the simulation accurately reflects the chronological sequence of events and the dynamic evolution of the logistics operation.

### 6. Consider implementing an event-driven architecture

To manage the complex and asynchronous nature of dynamic routing simulations, an event-driven architecture is highly beneficial. This approach allows the simulation to react efficiently to changes without continuously polling for updates.

#### Core Principles:
*   **Events:** Everything that happens in the simulation is treated as an event (e.g., `OrderArrived`, `VehicleDeparted`, `VehicleArrived`, `TrafficUpdate`, `DriverFinishedShift`).
*   **Event Queue:** A central component that stores scheduled events, ordered by their occurrence time. The simulation clock advances by processing events from this queue.
*   **Event Handlers:** Specific functions or modules designed to react to and process particular types of events.

#### Benefits of Event-Driven Architecture:
*   **Efficiency:** The simulation only performs computations when an event occurs, rather than at fixed, potentially idle, time steps. This can significantly reduce computational overhead, especially in sparsely active scenarios.
*   **Flexibility:** Easily add new types of events and corresponding handlers without overhacomplicating the core simulation logic.
*   **Realism:** Naturally models real-world logistics operations where events (like new orders or traffic incidents) occur asynchronously.
*   **Modularity:** Promotes a modular design where different components (e.g., vehicle movement, order management, routing logic) interact primarily through events.

#### Key Components:
*   **`Event` Class:** A base class or structure for all events, containing at least an `event_time` and `event_type`.
*   **`SimulationEngine`:** The main orchestrator that manages the event queue, advances the simulation clock, and dispatches events to appropriate handlers.
*   **`EventHandler` Classes/Functions:** Specific logic for handling each event type. For example, an `OrderArrivalHandler` would add a new order to the system and potentially trigger a routing re-optimization.

#### Example Flow:
1.  Initialize the `Event` queue with initial events (e.g., first set of orders, vehicle departures from depot).
2.  The `SimulationEngine` retrieves the next event from the queue (the one with the earliest `event_time`).
3.  The simulation clock is advanced to `event_time`.
4.  The `SimulationEngine` dispatches the event to its registered handler.
5.  The handler processes the event, updates the state of the simulation, and may schedule new future events (e.g., `VehicleArrived` event after `VehicleDeparted`).
6.  Repeat until the `end_time` is reached or the event queue is empty.

This architecture is particularly well-suited for dynamic routing where real-time information and unexpected occurrences necessitate agile responses.

### 7. Decide on simulation library vs. custom framework

Before diving into implementation, a fundamental decision needs to be made regarding the underlying framework for the simulation: whether to use an existing simulation library or to build a custom solution from scratch.

#### Options:

*   **Utilize Existing Simulation Libraries (e.g., SimPy, Mesa, Anylogic, ARENA):**
    *   **Description:** These libraries provide pre-built constructs and methodologies for discrete-event simulation (like SimPy) or agent-based modeling (like Mesa), handling aspects such as time progression, event scheduling, and agent interactions.
    *   **Pros:**
        *   **Accelerated Development:** Many common simulation tasks are already implemented.
        *   **Robustness:** Libraries are often well-tested and maintained.
        *   **Community Support:** Access to documentation, examples, and community forums.
        *   **Focus on Logic:** Allows developers to concentrate more on the specific routing logic rather than the underlying simulation mechanics.
        *   **Proven Methodologies:** Often implement established simulation principles effectively.
    *   **Cons:**
        *   **Learning Curve:** Requires understanding the library's specific API and paradigms.
        *   **Less Flexibility:** May impose certain architectural constraints or be less adaptable to highly unique requirements.
        *   **Overhead:** Might include functionalities not needed for the specific problem, potentially adding unnecessary complexity or performance overhead.

*   **Build a Custom Framework from Scratch:**
    *   **Description:** Designing and implementing all simulation components (e.g., event queue, time management, object models for vehicles/orders) without relying on a dedicated simulation library.
    *   **Pros:**
        *   **Maximum Flexibility:** Complete control over every aspect of the simulation's design and implementation.
        *   **Optimization:** Can be highly tailored and optimized for the specific problem, potentially leading to better performance for very niche requirements.
        *   **Deeper Understanding:** Provides a thorough understanding of simulation mechanics.
    *   **Cons:**
        *   **Time-Consuming:** Significantly longer development time due to re-implementing basic simulation functionalities.
        *   **Higher Risk of Bugs:** Increased likelihood of errors in fundamental simulation logic.
        *   **Maintenance Burden:** Ongoing maintenance and debugging are entirely the responsibility of the development team.
        *   **Lack of Best Practices:** Without adherence to established patterns, the design might not be as robust or scalable.

#### Decision Criteria:

*   **Complexity of the Problem:** For highly complex and unique dynamic routing scenarios, a custom framework might be necessary, but for more standard problems, a library is often sufficient.
*   **Development Time and Resources:** Libraries are generally preferred for projects with tight deadlines or limited resources.
*   **Required Level of Detail/Accuracy:** If extreme precision or very specific custom behaviors are needed, a custom approach offers more control.
*   **Team Expertise:** The team's familiarity with existing simulation libraries versus general programming skills will influence the choice.

For this project, given the need for dynamic routing and integration with a prediction model, a **hybrid approach or leveraging a flexible discrete-event simulation library like SimPy** is likely suitable. It offers a good balance between speed of development and customization capability. The event-driven architecture discussed in Step 6 aligns well with such libraries.

````python
import datetime

class Location:
    def __init__(self, id, latitude, longitude, name=None):
        self.id = id
        self.latitude = latitude
        self.longitude = longitude
        self.name = name

    def __repr__(self):
        return f"Location(ID: {self.id}, Lat: {self.latitude}, Lon: {self.longitude}, Name: {self.name})"

class Order:
    def __init__(self, id, origin_location_id, destination_location_id, quantity, time_window_start, time_window_end, service_time_seconds):
        self.id = id
        self.origin_location_id = origin_location_id
        self.destination_location_id = destination_location_id
        self.quantity = quantity
        self.time_window_start = time_window_start # datetime object
        self.time_window_end = time_window_end   # datetime object
        self.service_time_seconds = service_time_seconds # time spent at destination

    def __repr__(self):
        return f"Order(ID: {self.id}, From: {self.origin_location_id}, To: {self.destination_location_id}, Qty: {self.quantity}, TW: {self.time_window_start.strftime('%H:%M')}-{self.time_window_end.strftime('%H:%M')})"

class Vehicle:
    def __init__(self, id, capacity, current_location_id, current_time, assigned_orders=None, current_route=None):
        self.id = id
        self.capacity = capacity
        self.current_location_id = current_location_id
        self.current_time = current_time # datetime object, vehicle's current time
        self.assigned_orders = assigned_orders if assigned_orders is not None else [] # List of Order objects
        self.current_route = current_route if current_route is not None else [] # List of Location IDs in order
        self.current_load = 0

    def add_order_to_route(self, order, delivery_location_id):
        # For simplicity, we assume one order corresponds to one delivery location
        self.assigned_orders.append(order)
        self.current_route.append(delivery_location_id)
        self.current_load += order.quantity

    def __repr__(self):
        return f"Vehicle(ID: {self.id}, Capacity: {self.capacity}, Current Loc: {self.current_location_id}, Load: {self.current_load}, Orders: {len(self.assigned_orders)})"

print("VRP component classes (Location, Order, Vehicle) defined successfully.")
````

The next instruction is to implement a function to calculate distances and travel times between any two locations. I will use the Haversine formula for initial distance calculation, which is suitable for geographical coordinates, and then derive a simple travel time from that.



````python
from math import radians, sin, cos, sqrt, atan2

def haversine_distance(loc1: Location, loc2: Location) -> float:
    """
    Calculate the distance between two geographical locations using the Haversine formula.
    Returns distance in kilometers.
    """
    R = 6371  # Radius of Earth in kilometers

    lat1, lon1 = radians(loc1.latitude), radians(loc1.longitude)
    lat2, lon2 = radians(loc2.latitude), radians(loc2.longitude)

    dlon = lon2 - lon1
    dlat = lat2 - lat1

    a = sin(dlat / 2)**2 + cos(lat1) * cos(lat2) * sin(dlon / 2)**2
    c = 2 * atan2(sqrt(a), sqrt(1 - a))

    distance = R * c
    return distance

def calculate_travel_time(loc1: Location, loc2: Location, avg_speed_kmh: float = 40) -> float:
    """
    Calculate the estimated travel time between two locations based on distance and average speed.
    Returns time in hours.
    """
    distance = haversine_distance(loc1, loc2)
    if avg_speed_kmh <= 0:
        return float('inf') # Avoid division by zero or non-positive speed
    travel_time_hours = distance / avg_speed_kmh
    return travel_time_hours

print("Distance and travel time calculation functions defined (haversine_distance, calculate_travel_time).")
````

I will use a greedy insertion heuristic, which prioritizes orders (by earliest time window) and attempts to insert them into existing vehicle routes at the position that incurs the minimum additional cost (travel time), while respecting vehicle capacity and time window constraints. This involves updating the Vehicle class for better route management and creating a helper function to calculate the temporal feasibility and cost of a proposed route.



````python
import datetime
from typing import List, Dict, Tuple
from math import radians, sin, cos, sqrt, atan2

# Assuming Location and Order classes are defined from previous steps
# Redefine Vehicle class for better route management to store planned orders in sequence
class Vehicle:
    def __init__(self, id: str, capacity: float, current_physical_location_id: str, current_time: datetime.datetime, depot_location_id: str):
        self.id = id
        self.capacity = capacity
        self.current_physical_location_id = current_physical_location_id # Where the vehicle physically is at current_time
        self.current_time = current_time # Actual time at current_physical_location_id
        self.depot_location_id = depot_location_id # The vehicle's home base, assumed to be its initial location
        self.planned_route_orders: List[Order] = [] # List of Order objects in sequence for the planned route
        self.current_load = 0.0 # Sum of quantities of planned_route_orders

    def add_order_to_planned_route(self, order: Order, insertion_pos: int):
        self.planned_route_orders.insert(insertion_pos, order)
        self.current_load += order.quantity

    def remove_order_from_planned_route(self, order: Order):
        self.planned_route_orders.remove(order)
        self.current_load -= order.quantity

    def __repr__(self):
        return f"Vehicle(ID: {self.id}, Cap: {self.capacity}, Loc: {self.current_physical_location_id}, Load: {self.current_load}, Orders: {len(self.planned_route_orders)})
Planned Route: {[order.id for order in self.planned_route_orders]}"

def calculate_temporal_feasibility_and_cost(
    vehicle: Vehicle,
    proposed_planned_route_orders: List[Order],
    all_locations: Dict[str, Location],
    avg_speed_kmh: float
) -> Tuple[float, bool]:
    """
    Calculates the total travel time for a proposed vehicle route and checks time window constraints.
    Returns (total_route_time_seconds, is_feasible).
    The route includes travel from the vehicle's current physical location to the first stop,
    then between all delivery stops, and finally back to the depot.
    """
    current_loc_id = vehicle.current_physical_location_id
    current_time_at_loc = vehicle.current_time
    total_route_time_seconds = 0.0
    is_feasible = True

    # Sequence of stops: (location_id, Order_object_if_delivery_stop_else_None)
    route_sequence = []
    for order in proposed_planned_route_orders:
        route_sequence.append((order.destination_location_id, order))

    # Add the final return to depot
    route_sequence.append((vehicle.depot_location_id, None))

    # Simulate the route step by step
    for i in range(len(route_sequence)):
        next_stop_loc_id, order_at_next_stop = route_sequence[i]

        # Travel from current_loc_id to next_stop_loc_id
        travel_time_hours = calculate_travel_time(all_locations[current_loc_id], all_locations[next_stop_loc_id], avg_speed_kmh)
        travel_time_seconds = travel_time_hours * 3600

        current_time_at_loc += datetime.timedelta(seconds=travel_time_seconds)
        total_route_time_seconds += travel_time_seconds

        # Process arrival at next_stop_loc_id (if it's a delivery)
        if order_at_next_stop: # If this is a delivery point for an order
            # Check time window for the order
            if current_time_at_loc < order_at_next_stop.time_window_start:
                # Arrived early, wait until time window starts
                wait_time_seconds = (order_at_next_stop.time_window_start - current_time_at_loc).total_seconds()
                current_time_at_loc = order_at_next_stop.time_window_start
                total_route_time_seconds += wait_time_seconds
            elif current_time_at_loc > order_at_next_stop.time_window_end:
                # Arrived late, constraint violated
                is_feasible = False
                break

            # Add service time for the order
            current_time_at_loc += datetime.timedelta(seconds=order_at_next_stop.service_time_seconds)
            total_route_time_seconds += order_at_next_stop.service_time_seconds

        current_loc_id = next_stop_loc_id # Move to the next location in the simulation

    return total_route_time_seconds, is_feasible

def greedy_insertion_routing(
    vehicles: List[Vehicle],
    pending_orders: List[Order],
    all_locations: Dict[str, Location],
    avg_speed_kmh: float = 40
) -> Tuple[List[Vehicle], List[Order]]:
    """
    Implements a greedy insertion heuristic for dynamic routing.
    Assigns pending orders to vehicles by finding the insertion that
    results in the minimum additional route cost while respecting constraints.
    Orders are sorted by their time window start to prioritize earlier deliveries.

    Args:
        vehicles (List[Vehicle]): A list of available Vehicle objects.
        pending_orders (List[Order]): A list of Order objects waiting to be assigned.
        all_locations (Dict[str, Location]): A dictionary of all known Location objects by ID.
        avg_speed_kmh (float): Average speed for travel time calculations.

    Returns:
        Tuple[List[Vehicle], List[Order]]: The updated list of vehicles with assigned routes
                                          and a list of orders that could not be assigned.
    """
    unassigned_orders = list(pending_orders)

    # Sort orders by their time window start to prioritize earlier deliveries.
    unassigned_orders.sort(key=lambda o: o.time_window_start)

    # Loop until no more orders can be assigned in this iteration
    # or all orders are assigned.
    while unassigned_orders:
        order_to_assign = unassigned_orders[0] # Pick the next highest priority unassigned order

        best_vehicle_for_order = None
        best_insertion_pos_for_order = -1
        min_additional_cost_for_order = float('inf')

        for vehicle in vehicles:
            # Capacity check: does adding this order exceed vehicle capacity?
            if vehicle.current_load + order_to_assign.quantity > vehicle.capacity:
                continue

            # Calculate the current cost of this vehicle's route BEFORE any new insertion
            current_route_cost, current_route_feasible = calculate_temporal_feasibility_and_cost(
                vehicle,
                vehicle.planned_route_orders,
                all_locations,
                avg_speed_kmh
            )

            if not current_route_feasible: # If the vehicle's current route is already infeasible, skip it
                continue

            # Try inserting the order at all possible positions in the vehicle's current planned route
            # Positions range from 0 (start) to len(planned_route_orders) (end)
            for pos in range(len(vehicle.planned_route_orders) + 1):
                # Create a temporary list of orders for the proposed route including the new order
                proposed_route_orders_temp = vehicle.planned_route_orders[:pos] + [order_to_assign] + vehicle.planned_route_orders[pos:]

                # Calculate the cost and check feasibility for this proposed route
                new_total_cost, is_feasible = calculate_temporal_feasibility_and_cost(
                    vehicle,
                    proposed_route_orders_temp,
                    all_locations,
                    avg_speed_kmh
                )

                if is_feasible:
                    # Calculate additional cost compared to the vehicle's current route cost
                    # Handle initial empty route: current_route_cost would be 0 if vehicle.planned_route_orders is empty
                    additional_cost = new_total_cost - current_route_cost

                    if additional_cost < min_additional_cost_for_order:
                        min_additional_cost_for_order = additional_cost
                        best_vehicle_for_order = vehicle
                        best_insertion_pos_for_order = pos

        # After checking all vehicles and all positions for the current order_to_assign
        if best_vehicle_for_order is not None and min_additional_cost_for_order < float('inf'):
            best_vehicle_for_order.add_order_to_planned_route(order_to_assign, best_insertion_pos_for_order)
            unassigned_orders.pop(0) # Remove the assigned order from the front of the list
        else:
            # If the current order cannot be assigned to any vehicle, it remains unassigned
            # and we break the loop to move on to evaluating remaining unassigned orders.
            break

    return vehicles, unassigned_orders

print("Updated Vehicle class and greedy_insertion_routing algorithm defined, along with feasibility and cost calculation helper.")
````

````python
import datetime
from typing import List, Dict, Tuple
from math import radians, sin, cos, sqrt, atan2

# Assuming Location and Order classes are defined from previous steps
# Redefine Vehicle class for better route management to store planned orders in sequence
class Vehicle:
    def __init__(self, id: str, capacity: float, current_physical_location_id: str, current_time: datetime.datetime, depot_location_id: str):
        self.id = id
        self.capacity = capacity
        self.current_physical_location_id = current_physical_location_id # Where the vehicle physically is at current_time
        self.current_time = current_time # Actual time at current_physical_location_id
        self.depot_location_id = depot_location_id # The vehicle's home base, assumed to be its initial location
        self.planned_route_orders: List[Order] = [] # List of Order objects in sequence for the planned route
        self.current_load = 0.0 # Sum of quantities of planned_route_orders

    def add_order_to_planned_route(self, order: Order, insertion_pos: int):
        self.planned_route_orders.insert(insertion_pos, order)
        self.current_load += order.quantity

    def remove_order_from_planned_route(self, order: Order):
        self.planned_route_orders.remove(order)
        self.current_load -= order.quantity

    def __repr__(self):
        return f"Vehicle(ID: {self.id}, Cap: {self.capacity}, Loc: {self.current_physical_location_id}, Load: {self.current_load}, Orders: {len(self.planned_route_orders)}, Planned Route: {[order.id for order in self.planned_route_orders]})"

def calculate_travel_time(loc1: 'Location', loc2: 'Location', avg_speed_kmh: float = 40) -> float:
    """
    Calculate the estimated travel time between two locations based on distance and average speed.
    Returns time in hours.
    """
    # Re-using haversine_distance from the previous step
    R = 6371  # Radius of Earth in kilometers

    lat1, lon1 = radians(loc1.latitude), radians(loc1.longitude)
    lat2, lon2 = radians(loc2.latitude), radians(loc2.longitude)

    dlon = lon2 - lon1
    dlat = lat2 - lat1

    a = sin(dlat / 2)**2 + cos(lat1) * cos(lat2) * sin(dlon / 2)**2
    c = 2 * atan2(sqrt(a), sqrt(1 - a))

    distance = R * c

    if avg_speed_kmh <= 0:
        return float('inf') # Avoid division by zero or non-positive speed
    travel_time_hours = distance / avg_speed_kmh
    return travel_time_hours

def calculate_temporal_feasibility_and_cost(
    vehicle: Vehicle,
    proposed_planned_route_orders: List['Order'],
    all_locations: Dict[str, 'Location'],
    avg_speed_kmh: float
) -> Tuple[float, bool]:
    """
    Calculates the total travel time for a proposed vehicle route and checks time window constraints.
    Returns (total_route_time_seconds, is_feasible).
    The route includes travel from the vehicle's current physical location to the first stop,
    then between all delivery stops, and finally back to the depot.
    """
    current_loc_id = vehicle.current_physical_location_id
    current_time_at_loc = vehicle.current_time
    total_route_time_seconds = 0.0
    is_feasible = True

    # Sequence of stops: (location_id, Order_object_if_delivery_stop_else_None)
    route_sequence = []
    for order in proposed_planned_route_orders:
        route_sequence.append((order.destination_location_id, order))

    # Add the final return to depot
    route_sequence.append((vehicle.depot_location_id, None))

    # Simulate the route step by step
    for i in range(len(route_sequence)):
        next_stop_loc_id, order_at_next_stop = route_sequence[i]

        # Travel from current_loc_id to next_stop_loc_id
        travel_time_hours = calculate_travel_time(all_locations[current_loc_id], all_locations[next_stop_loc_id], avg_speed_kmh)
        travel_time_seconds = travel_time_hours * 3600

        current_time_at_loc += datetime.timedelta(seconds=travel_time_seconds)
        total_route_time_seconds += travel_time_seconds

        # Process arrival at next_stop_loc_id (if it's a delivery)
        if order_at_next_stop: # If this is a delivery point for an order
            # Check time window for the order
            if current_time_at_loc < order_at_next_stop.time_window_start:
                # Arrived early, wait until time window starts
                wait_time_seconds = (order_at_next_stop.time_window_start - current_time_at_loc).total_seconds()
                current_time_at_loc = order_at_next_stop.time_window_start
                total_route_time_seconds += wait_time_seconds
            elif current_time_at_loc > order_at_next_stop.time_window_end:
                # Arrived late, constraint violated
                is_feasible = False
                break

            # Add service time for the order
            current_time_at_loc += datetime.timedelta(seconds=order_at_next_stop.service_time_seconds)
            total_route_time_seconds += order_at_next_stop.service_time_seconds

        current_loc_id = next_stop_loc_id # Move to the next location in the simulation

    return total_route_time_seconds, is_feasible

def greedy_insertion_routing(
    vehicles: List[Vehicle],
    pending_orders: List['Order'],
    all_locations: Dict[str, 'Location'],
    avg_speed_kmh: float = 40
) -> Tuple[List[Vehicle], List['Order']]:
    """
    Implements a greedy insertion heuristic for dynamic routing.
    Assigns pending orders to vehicles by finding the insertion that
    results in the minimum additional route cost while respecting constraints.
    Orders are sorted by their time window start to prioritize earlier deliveries.

    Args:
        vehicles (List[Vehicle]): A list of available Vehicle objects.
        pending_orders (List[Order]): A list of Order objects waiting to be assigned.
        all_locations (Dict[str, Location]): A dictionary of all known Location objects by ID.
        avg_speed_kmh (float): Average speed for travel time calculations.

    Returns:
        Tuple[List[Vehicle], List[Order]]: The updated list of vehicles with assigned routes
                                          and a list of orders that could not be assigned.
    """
    unassigned_orders = list(pending_orders)

    # Sort orders by their time window start to prioritize earlier deliveries.
    unassigned_orders.sort(key=lambda o: o.time_window_start)

    # Loop until no more orders can be assigned in this iteration
    # or all orders are assigned.
    while unassigned_orders:
        order_to_assign = unassigned_orders[0] # Pick the next highest priority unassigned order

        best_vehicle_for_order = None
        best_insertion_pos_for_order = -1
        min_additional_cost_for_order = float('inf')

        for vehicle in vehicles:
            # Capacity check: does adding this order exceed vehicle capacity?
            if vehicle.current_load + order_to_assign.quantity > vehicle.capacity:
                continue

            # Calculate the current cost of this vehicle's route BEFORE any new insertion
            current_route_cost, current_route_feasible = calculate_temporal_feasibility_and_cost(
                vehicle,
                vehicle.planned_route_orders,
                all_locations,
                avg_speed_kmh
            )

            if not current_route_feasible: # If the vehicle's current route is already infeasible, skip it
                continue

            # Try inserting the order at all possible positions in the vehicle's current planned route
            # Positions range from 0 (start) to len(planned_route_orders) (end)
            for pos in range(len(vehicle.planned_route_orders) + 1):
                # Create a temporary list of orders for the proposed route including the new order
                proposed_route_orders_temp = vehicle.planned_route_orders[:pos] + [order_to_assign] + vehicle.planned_route_orders[pos:]

                # Calculate the cost and check feasibility for this proposed route
                new_total_cost, is_feasible = calculate_temporal_feasibility_and_cost(
                    vehicle,
                    proposed_route_orders_temp,
                    all_locations,
                    avg_speed_kmh
                )

                if is_feasible:
                    # Calculate additional cost compared to the vehicle's current route cost
                    # Handle initial empty route: current_route_cost would be 0 if vehicle.planned_route_orders is empty
                    additional_cost = new_total_cost - current_route_cost

                    if additional_cost < min_additional_cost_for_order:
                        min_additional_cost_for_order = additional_cost
                        best_vehicle_for_order = vehicle
                        best_insertion_pos_for_order = pos

        # After checking all vehicles and all positions for the current order_to_assign
        if best_vehicle_for_order is not None and min_additional_cost_for_order < float('inf'):
            best_vehicle_for_order.add_order_to_planned_route(order_to_assign, best_insertion_pos_for_order)
            unassigned_orders.pop(0) # Remove the assigned order from the front of the list
        else:
            # If the current order cannot be assigned to any vehicle, it remains unassigned
            # and we break the loop to move on to evaluating remaining unassigned orders.
            break

    return vehicles, unassigned_orders

print("Updated Vehicle class and greedy_insertion_routing algorithm defined, along with feasibility and cost calculation helper.")
````

## Define a clear baseline scenario for comparison

````python
from typing import List, Dict, Tuple

def baseline_routing_simple_assignment(
    vehicles: List[Vehicle],
    pending_orders: List[Order],
    all_locations: Dict[str, Location],
    avg_speed_kmh: float = 40
) -> Tuple[List[Vehicle], List[Order]]:
    """
    Implements a simple baseline routing heuristic. Orders are assigned sequentially
    to the first available vehicle that can accommodate them and maintain route feasibility.

    Args:
        vehicles (List[Vehicle]): A list of available Vehicle objects.
        pending_orders (List[Order]): A list of Order objects waiting to be assigned.
        all_locations (Dict[str, Location]): A dictionary of all known Location objects by ID.
        avg_speed_kmh (float): Average speed for travel time calculations.

    Returns:
        Tuple[List[Vehicle], List[Order]]: The updated list of vehicles with assigned routes
                                          and a list of orders that could not be assigned.
    """
    unassigned_orders = []
    assigned_orders_count = 0

    for order_to_assign in pending_orders:
        order_assigned_in_this_pass = False
        for vehicle in vehicles:
            # 1. Capacity check
            if vehicle.current_load + order_to_assign.quantity > vehicle.capacity:
                continue # Vehicle cannot carry this order

            # 2. Create a temporary proposed route by appending to the end
            # The insertion position is simply the end of the current planned orders
            proposed_route_orders_temp = vehicle.planned_route_orders + [order_to_assign]

            # 3. Check temporal feasibility and cost for this proposed route
            _, is_feasible = calculate_temporal_feasibility_and_cost(
                vehicle,
                proposed_route_orders_temp,
                all_locations,
                avg_speed_kmh
            )

            # 4. If feasible, assign the order
            if is_feasible:
                vehicle.add_order_to_planned_route(order_to_assign, len(vehicle.planned_route_orders)) # Insert at end
                assigned_orders_count += 1
                order_assigned_in_this_pass = True
                break # Move to the next order, this vehicle got the order

        if not order_assigned_in_this_pass:
            unassigned_orders.append(order_to_assign)

    print(f"Assigned {assigned_orders_count} orders using baseline routing.")
    return vehicles, unassigned_orders

print("Baseline routing (simple assignment) function defined.")
````

    *   **Description:** The logistics network will be modeled as a graph where intersections and key locations are represented as **nodes**, and roads connecting them are represented as **edges**.

            *   **Distance:** Actual physical length of the road segment.
            *   **Base Travel Time:** Travel time under free-flow conditions (calculated from distance and speed limit).
            *   **Speed Limits:** Associated speed limits for each road segment.
            *   **Road Type:** (e.g., highway, urban road, residential street) which can influence speed and traffic patterns.
            *   **One-Way Restrictions:** Indicate if a road segment is one-way.
            *   **Traffic Conditions:** Capacity or congestion levels (for dynamic updates).

## Collect key performance metrics from both simulation runs

### Subtask:
Collect key performance metrics, such as total travel time, total distance, and the number of unassigned orders, from both the greedy and baseline simulation environments.


**Reasoning**:
I need to collect key performance metrics from both the greedy and baseline simulation environments. This involves iterating through each vehicle in each environment to calculate total travel time, total assigned orders, and then recording the total unassigned orders. I will store these metrics in dictionaries for each scenario.



````python
import math

# --- Function to calculate total route metrics for a given simulation environment ---
def get_scenario_metrics(sim_env: SimulationEnvironment) -> Dict[str, float]:
    total_travel_time_seconds = 0.0
    total_assigned_orders = 0

    for vehicle in sim_env.vehicles:
        # Calculate the total travel time for the vehicle's currently planned route
        # This function returns (total_route_time_seconds, is_feasible).
        # We only care about the time here.
        # If a vehicle has no assigned orders, its planned route is empty, and the cost will be 0.
        route_time_seconds, _ = calculate_temporal_feasibility_and_cost(
            vehicle,
            vehicle.planned_route_orders, # Pass the current planned orders of the vehicle
            sim_env.all_locations,
            sim_env.avg_speed_kmh
        )
        total_travel_time_seconds += route_time_seconds
        total_assigned_orders += len(vehicle.planned_route_orders)

    total_unassigned_orders = len(sim_env.pending_orders)

    return {
        "total_travel_time_seconds": total_travel_time_seconds,
        "total_travel_time_hours": total_travel_time_seconds / 3600,
        "total_assigned_orders": float(total_assigned_orders),
        "total_unassigned_orders": float(total_unassigned_orders)
    }

# --- Collect metrics for both scenarios ---
metrics_greedy = get_scenario_metrics(sim_env_greedy)
metrics_baseline = get_scenario_metrics(sim_env_baseline)

print("\n--- Collected Metrics (Greedy Scenario) ---")
for key, value in metrics_greedy.items():
    print(f"{key}: {value:.2f}")

print("\n--- Collected Metrics (Baseline Scenario) ---")
for key, value in metrics_baseline.items():
    print(f"{key}: {value:.2f}")
````
